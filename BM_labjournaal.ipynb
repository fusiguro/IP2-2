```python
#Data processing and analysis:
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

def function(x, a, b):
    return a*x+b

log_I = np.log(I)

val, cov = curve_fit(function, U_diode, log_I)

plt.figure()
plt.semilogy(U_diode, I, '.')
plt.plot(U_diode, function(U_diode, *val), 'r--')
plt.xlabel("U_diode (V)")
plt.ylabel("I_diode (A)")
plt.title("(U, I)-karakteristiek")
plt.grid()
plt.show()
```

```python
#Data processing and analysis:
# Constantes
q = 1.602e-19
T = 293.15
n = 2.0

# Logaritme van de stroom
log_I = np.log10(I)

a, b = val

# Berekening Boltzmannconstante
k_B = q * np.log10(np.e) / (n * T * a)

# Onzekerheid
da = np.sqrt(cov[0,0])
dk = k_B * da / abs(a)

print("Helling =", a)
print("Snijpunt =", b)
print("Boltzmannconstante =", k_B)
print("Onzekerheid =", dk)
```
