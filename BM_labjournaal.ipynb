**Method:**  Het experimenteel bepalen van de (I,V)-karakteristiek van een siliciumdiode. Met behulp van de meetgegevens wordt de constante van Boltzmann bepaald door theoretische diodidevergelijking te fitten aan de meetresultaten.

**Theory:**  de stroom door een hafgeleiderdiode hangt exponentieel af van de spanning over de diode. Dit verbanden wordt bescrheven door de diodevergelijking:

I_D = I_0(e^(q*U_D/n*k_B*T) - 1)

voor de spanningen over de diode groter dan ongeveer 0,1 V mag de -1 worden verwaarloosd. Hierdoor ontstaat:

I_D = I_0*e^(q*U_D/n*k_B*T)

door aan beide kanten de log te pakken met basis 10, krijgen we:

log(I_D) = log(I_0) + log(e^(q/n*k_B*T))*V_D

dit heeft de vorm van een rechte lijn:

y = ax + b

de helling van deze lijn kan worden gebruikt om de Boltzmannconstante te bepalen

k_B = q/n*T*a

hierbij geldt:
q: elementaire lading van een elektron
n: idealiteitsfactor van de diode
T: absolute temperatuur in Kelvin
a: helling van de gefitte lijn

**Independent variable:**  de voedingsspanning ($U_{bron}$)

**Dependent variable:**  spanning en stroomsterkte

**Controlled variablen:**  de spanning over de diode ($U_{diode}$)
 en de stroomsterkte door de diode ($I_{diode}$)

# Processing
**Description of processing of raw data into scientific evidence:**


```python
#Lineaire functie
def function(x, a, b):
    return a*x + b

#Constantes
q = 1.602176634e-19
n = 2.0
u_n = 0.1
T = 293.15

#Alleen meetpunten boven 0.1 V worden gebruikt
U_diode_mask = U_diode > 0.1
log_I = np.log(I[U_diode_mask])

val, cov = curve_fit(function, U_diode[U_diode_mask], log_I)

a, b = val

#Zorgt ervoor dat de gefitte lijn over het gehele domein wordt geplot
U_lijn = np.linspace(0, max(U_diode), 200)

plt.figure()
plt.semilogy(U_diode, I, '.', label = "Meetpunten")
plt.semilogy(U_lijn, np.exp(function(U_lijn, *val)), 'r--', label = "lineaire fit")
plt.xlabel("$U_{diode}$ (V)")
plt.ylabel("$I_{diode}$ (A)")
plt.title("(U,I)-karakteristiek")
plt.legend()
plt.grid()
plt.show()

#Boltzmannconstante
k_B = q/(n*T*a)
k_B_afwijking = [q/((n+u_n)*T*a), q/((n-u_n)*T*a)]

#Onzekerheid
d_a = np.sqrt(cov[0,0])
dk_B = k_B * d_a/a

#afwijking
k_B_actueel = 1.380649e-23
percentage_afwijking = np.abs(k_B_actueel - k_B)/k_B_actueel * 100

print("Boltzmannconstante (n = 1.9):", k_B_afwijking[0])
print("Boltzmannconstante (n = 2):", k_B)
print("Boltzmannconstante (n = 2.1):", k_B_afwijking[1])
print(f"Afwijkingspercentage: {percentage_afwijking}%" )
print("Onzekerheid:", dk_B)
```
